# Wordnet Lemmatizer
WordNet Lemmatizer and morphy() Function in NLTK
The WordNet Lemmatizer in NLTK is used to reduce words to their base (dictionary) form, known as a lemma. It is particularly useful in NLP for text normalization.

🔹 What is morphy()?
The morphy() function is an internal method of the WordNet Lemmatizer that tries to find the base form (lemma) of a given word without considering context. If no lemma is found, it returns None.

```python
## QandA, chatbots, text summarization
import nltk

nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS - Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize("going",pos='v')
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
for word in words:
  print(word+ "----->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize("goes",pos='v')
```

```python
lemmatizer.lemmatize("fairly", pos='v'),lemmatizer.lemmatize("sportingly")
```
